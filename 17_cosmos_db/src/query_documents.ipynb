```python
from azure.cosmos import CosmosClient
from constants import COSMOS_URL, COSMOS_KEY, DB_ID, CONTAINER_ID

# connect to the cosmos instance, db and container
client = CosmosClient(COSMOS_URL, credential=COSMOS_KEY)
database = client.get_database_client(DB_ID)
container = database.get_container_client(CONTAINER_ID)

# query without filter
print("List all documents:")
query = "SELECT * FROM c"

items = list(container.query_items(query=query, enable_cross_partition_query=True))
items
```

```python
for item in items:
    print(item["title"], item["year"])
```

```python
# query with filter
print("List document(s) in a certain year: ")
query = "SELECT * FROM c WHERE c.year < @year"

items = list(
    container.query_items(
        query=query,
        parameters=[{"name": "@year", "value": 2023}],
        enable_cross_partition_query=True,
    )
)

for item in items:
    print(item["title"], item["year"])
```
